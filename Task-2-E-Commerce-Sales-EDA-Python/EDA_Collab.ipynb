```python
df["Product"].value_counts().plot(kind="bar", figsize=(8,5))
```

```python
import matplotlib.pyplot as plt

df["Product"].value_counts().plot(kind="bar", figsize=(8,5))

plt.title("Number of Orders by Product")
plt.xlabel("Product")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)

plt.show()
```

```python
df["OrderStatus"].value_counts().plot(kind="bar", figsize=(8,5))

plt.title("Number of Orders by Status")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.show()
```

```python
df["PaymentMethod"].value_counts().plot(
    kind="pie",
    figsize=(6,6),
    autopct="%1.1f%%"
)

plt.title("Payment Method Distribution")
plt.ylabel("")

plt.show()
```

```python
df["TotalPrice"].plot(kind="hist", figsize=(8,5), bins=10)

plt.title("Distribution of Total Order Price")
plt.xlabel("Total Price")
plt.ylabel("Number of Orders")

plt.show()
```

```python
df["TotalPrice"].plot(kind="box", figsize=(6,4))

plt.title("Box Plot of Total Price")
plt.ylabel("Total Price")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="Blues")

plt.title("Correlation Heatmap")
plt.show()
```

**Business Insights**
